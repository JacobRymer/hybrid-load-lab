# Testing to see what data loads from intervals.icu

```python
import pandas as pd

import pandas as pd

act = pd.read_csv("../data/processed/activities.csv", parse_dates=["start_date"])
well = pd.read_csv("../data/processed/wellness.csv", parse_dates=["date"])
print(act.shape, well.shape)
act.columns.tolist()
```

```python
print((act["load"] == act["hr_load"]).groupby(act["group"]).mean())
print((act["load"] / act["trimp"]).groupby(act["group"]).median())
```
